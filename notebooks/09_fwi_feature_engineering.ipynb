```python
# Buat ulang input FWI dari noon_weather
fwi_input = noon_weather.dropna(
    subset=[
        "temperature_2m",
        "relative_humidity_2m",
        "wind_speed_10m",
        "precipitation_24h"
    ]
).copy()

# Pastikan tipe tanggal benar
fwi_input["local_date"] = pd.to_datetime(fwi_input["local_date"])

# Urutkan
fwi_input = (
    fwi_input
    .sort_values(["lat", "lon", "local_date"])
    .reset_index(drop=True)
)

print("FWI input shape:", fwi_input.shape)
print(
    fwi_input[
        ["local_date", "lat", "lon",
         "temperature_2m",
         "relative_humidity_2m",
         "wind_speed_10m",
         "precipitation_24h"]
    ].head()
)
```

```python
# ==========================================
# HITUNG CANADIAN FWI - KALIMANTAN
# ==========================================

import numpy as np
import pandas as pd
from cffdrs import cffwis


def to_float(x):
    """Convert numpy scalar/array menjadi float."""
    return float(np.asarray(x).squeeze())


fwi_results = []

# Sequential calculation harus dilakukan per weather grid
for (lat, lon), group in fwi_input.groupby(["lat", "lon"], sort=False):

    group = group.sort_values("local_date").copy()

    # Initial FWI codes
    ffmc_prev = 85.0
    dmc_prev = 6.0
    dc_prev = 15.0

    for _, row in group.iterrows():

        temp = float(row["temperature_2m"])
        rh = float(row["relative_humidity_2m"])
        wind = float(row["wind_speed_10m"])
        precip = float(row["precipitation_24h"])

        month = pd.Timestamp(row["local_date"]).month

        # -------------------------
        # FFMC
        # -------------------------
        ffmc = to_float(
            cffwis.dailyFFMC(
                ffmc0=ffmc_prev,
                temp=temp,
                rh=rh,
                wind=wind,
                precip=precip
            )
        )

        # -------------------------
        # DMC
        # Latitude adjustment
        # -------------------------
        dmc = to_float(
            cffwis.dailyDMC(
                dmc0=dmc_prev,
                temp=temp,
                rh=rh,
                precip=precip,
                month=month,
                lat=lat,
                lat_adjust=True
            )
        )

        # -------------------------
        # DC
        # Latitude adjustment
        # -------------------------
        dc = to_float(
            cffwis.dailyDC(
                dc0=dc_prev,
                temp=temp,
                precip=precip,
                month=month,
                lat=lat,
                lat_adjust=True
            )
        )

        # -------------------------
        # ISI
        # -------------------------
        isi = to_float(
            cffwis.dailyISI(
                wind=wind,
                ffmc=ffmc
            )
        )

        # -------------------------
        # BUI
        # -------------------------
        bui = to_float(
            cffwis.dailyBUI(
                dmc=dmc,
                dc=dc
            )
        )

        # -------------------------
        # FWI
        # -------------------------
        fwi = to_float(
            cffwis.dailyFWI(
                isi=isi,
                bui=bui
            )
        )

        fwi_results.append({
            "date": pd.Timestamp(row["local_date"]).date(),
            "lat": lat,
            "lon": lon,
            "ffmc": ffmc,
            "dmc": dmc,
            "dc": dc,
            "isi": isi,
            "bui": bui,
            "fwi": fwi
        })

        # Update state untuk hari berikutnya
        ffmc_prev = ffmc
        dmc_prev = dmc
        dc_prev = dc


# Convert ke DataFrame
fwi_daily = pd.DataFrame(fwi_results)

print("FWI daily shape:", fwi_daily.shape)
print()
print(fwi_daily.head())
```

```python
# ==========================================
# FINAL EXPERIMENT:
# LOGISTIC REGRESSION + FIRE HISTORY + FWI
# ==========================================

import numpy as np
import pandas as pd

from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    average_precision_score,
    precision_score,
    recall_score,
    f1_score,
    accuracy_score,
    confusion_matrix
)

# ------------------------------------------
# 1. Feature definitions
# ------------------------------------------

fire_features = [
    "fire_count_lag1",
    "fire_count_lag3",
    "fire_count_lag7",
    "frp_sum_lag1",
    "frp_sum_lag3",
    "frp_sum_lag7",
    "fire_count_roll3",
    "fire_count_roll7",
    "frp_sum_roll3",
    "frp_sum_roll7"
]

fwi_features = [
    "ffmc",
    "dmc",
    "dc",
    "isi",
    "bui",
    "fwi"
]

features_fwi = fire_features + fwi_features
target = "target_fire_active_t1"

print("Number of features:", len(features_fwi))


# ------------------------------------------
# 2. Pastikan tanggal benar
# ------------------------------------------

df_fwi["date"] = pd.to_datetime(df_fwi["date"])


# ------------------------------------------
# 3. Temporal split
# ------------------------------------------

train_fwi = df_fwi[
    df_fwi["date"] < "2026-08-01"
].copy()

val_fwi = df_fwi[
    (df_fwi["date"] >= "2026-08-01") &
    (df_fwi["date"] < "2026-09-01")
].copy()

test_fwi = df_fwi[
    df_fwi["date"] >= "2026-09-01"
].copy()


# ------------------------------------------
# 4. X dan y
# ------------------------------------------

X_train = train_fwi[features_fwi]
y_train = train_fwi[target]

X_val = val_fwi[features_fwi]
y_val = val_fwi[target]

X_test = test_fwi[features_fwi]
y_test = test_fwi[target]


print("\nDataset:")
print("Train:", X_train.shape)
print("Validation:", X_val.shape)
print("Test:", X_test.shape)


# ------------------------------------------
# 5. Check missing values
# ------------------------------------------

print("\nMissing values:")
print(
    pd.concat([
        X_train.isna().sum(),
        X_val.isna().sum(),
        X_test.isna().sum()
    ], axis=1).max(axis=1)
)


# ------------------------------------------
# 6. Standardization
#    FIT ONLY ON TRAIN
# ------------------------------------------

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)
X_test_scaled = scaler.transform(X_test)


# ------------------------------------------
# 7. Logistic Regression
# ------------------------------------------

lr_fwi = LogisticRegression(
    max_iter=2000,
    random_state=42
)

lr_fwi.fit(
    X_train_scaled,
    y_train
)


# ------------------------------------------
# 8. Validation probability
# ------------------------------------------

val_prob = lr_fwi.predict_proba(
    X_val_scaled
)[:, 1]


# ------------------------------------------
# 9. PR-AUC
# ------------------------------------------

pr_auc_fwi = average_precision_score(
    y_val,
    val_prob
)

print("\n================================")
print("FWI VALIDATION RESULT")
print("================================")
print("PR-AUC:", round(pr_auc_fwi, 6))
```

```python
# ==========================================
# FINAL AQ EXPERIMENT:
# FIRE HISTORY + RAW WEATHER + AOD + HCHO
# ==========================================

from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import average_precision_score

fire_features = [
    "fire_count_lag1",
    "fire_count_lag3",
    "fire_count_lag7",
    "frp_sum_lag1",
    "frp_sum_lag3",
    "frp_sum_lag7",
    "fire_count_roll3",
    "fire_count_roll7",
    "frp_sum_roll3",
    "frp_sum_roll7"
]

weather_features = [
    "temperature_mean",
    "temperature_max",
    "relative_humidity_mean",
    "precipitation_sum",
    "boundary_layer_height_mean",
    "vapour_pressure_deficit_mean",
    "wind_speed_mean",
    "wind_speed_max",
    "wind_u_mean",
    "wind_v_mean"
]

aq_features = [
    "aerosol_optical_depth_mean",
    "formaldehyde_mean"
]

features_aod_hcho = (
    fire_features +
    weather_features +
    aq_features
)

target = "target_fire_active_t1"

print("Total features:", len(features_aod_hcho))
print(features_aod_hcho)
```

```python
train_aq = df_fwi[
    df_fwi["date"] < "2026-08-01"
].copy()

val_aq = df_fwi[
    (df_fwi["date"] >= "2026-08-01") &
    (df_fwi["date"] < "2026-09-01")
].copy()

test_aq = df_fwi[
    df_fwi["date"] >= "2026-09-01"
].copy()

X_train_aq = train_aq[features_aod_hcho]
y_train_aq = train_aq[target]

X_val_aq = val_aq[features_aod_hcho]
y_val_aq = val_aq[target]

X_test_aq = test_aq[features_aod_hcho]
y_test_aq = test_aq[target]

print("Train:", X_train_aq.shape)
print("Validation:", X_val_aq.shape)
print("Test:", X_test_aq.shape)

print("\nMissing:")
print(
    X_train_aq.isna().sum() |
    X_val_aq.isna().sum() |
    X_test_aq.isna().sum()
)
```

```python
scaler_aq = StandardScaler()

X_train_aq_scaled = scaler_aq.fit_transform(X_train_aq)
X_val_aq_scaled = scaler_aq.transform(X_val_aq)
X_test_aq_scaled = scaler_aq.transform(X_test_aq)

lr_aod_hcho = LogisticRegression(
    max_iter=2000,
    random_state=42
)

lr_aod_hcho.fit(
    X_train_aq_scaled,
    y_train_aq
)

val_prob_aq = lr_aod_hcho.predict_proba(
    X_val_aq_scaled
)[:, 1]

pr_auc_aod_hcho = average_precision_score(
    y_val_aq,
    val_prob_aq
)

print("\n========================================")
print("AOD + HCHO VALIDATION RESULT")
print("========================================")
print("PR-AUC:", round(pr_auc_aod_hcho, 6))

print("\nComparison:")
print("Baseline LR + Fire + Weather :", 0.901149)
print("LR + Fire + Weather + AOD/HCHO:", round(pr_auc_aod_hcho, 6))
print(
    "Difference:",
    round(pr_auc_aod_hcho - 0.901149, 6)
)
```
