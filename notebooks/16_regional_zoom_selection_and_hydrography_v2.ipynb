## Penting: zoom visual ≠ resolusi model

Memperbesar gambar 0.75° tidak menambah detail spasial.

Karena itu notebook ini melakukan dua hal sekaligus:

```text
Kalimantan-wide raw FIRMS
        ↓
data-driven regional selection
        ↓
regional hydrography inspection
        ↓
0.25° / 0.10° fine-grid preview
```

---

## Data selection principle

Subregion **tidak dipilih karena performa model**.

Kandidat dipilih dari raw FIRMS menggunakan:

1. jumlah 3-day Emerging positives pada preview grid 0.25°;
2. jumlah active days;
3. spatial spread / active fine-grid cells;
4. jumlah deteksi FIRMS.

Hydrography dipakai sebagai **descriptive / realism check**, bukan untuk memilih area yang “membuat model terlihat bagus”.

---

## Hydrography caveat

Notebook ini memakai **Natural Earth 1:10m** untuk screening visual:

- land;
- country / province boundaries;
- lakes;
- rivers + lake centerlines.

Layer ini bagus untuk regional context, tetapi **bukan exhaustive high-resolution water mask**. Sungai kecil, kanal, rawa, dan badan air kecil dapat tidak muncul.

Jika subregion sudah dipilih, Notebook 17 sebaiknya memakai water mask yang lebih detail (misalnya JRC Global Surface Water / HydroSHEDS-compatible source) untuk final fine-grid construction.

---

# PART A — Load raw FIRMS detections

## 3. Locate the original Kalimantan FIRMS daily folder

The packaged dataset documents daily files as:

```text
data/kalimantan/fire_firms/daily/fires_<date>.csv
```

Each row is one VIIRS-SNPP thermal-anomaly detection with `lat`, `lon`, and `frp`.

````python
# 3. Discover KALIMANTAN raw FIRMS folder
# Prefer hourly because the packaged documentation identifies it as the
# more complete FIRMS record.

from pathlib import Path

# ============================================================
# OPTIONAL MANUAL OVERRIDE
# ============================================================
# If auto-detection does not find your KARHUT_Data_Share folder,
# set this to its root path, for example:
#
# DATASET_ROOT = Path(
#     r"C:\Users\Asus\Documents\KARHUT_Data_Share"
# )
#
# Otherwise leave it as None.
# ============================================================

DATASET_ROOT = None


def has_firms_files(folder):
    return (
        folder.exists()
        and folder.is_dir()
        and any(folder.glob("fires_*.csv"))
    )


def is_kalimantan_path(path):
    parts_lower = [
        str(part).lower()
        for part in Path(path).parts
    ]
    return "kalimantan" in parts_lower


def find_kalimantan_firms_folder(dataset_root=None):
    search_roots = []

    if dataset_root is not None:
        search_roots.append(
            Path(dataset_root)
        )

    search_roots.extend([
        PROJECT_ROOT,
        PROJECT_ROOT.parent,
        Path.home() / "Documents",
        Path.home() / "Downloads",
    ])

    # Remove duplicate / non-existing roots.
    unique_roots = []
    seen = set()

    for root in search_roots:
        if not root.exists():
            continue

        try:
            resolved = root.resolve()
        except Exception:
            resolved = root

        key = str(resolved).lower()

        if key not in seen:
            unique_roots.append(root)
            seen.add(key)

    # --------------------------------------------------------
    # 1. Try expected direct locations first.
    # --------------------------------------------------------
    direct_candidates = []

    for root in unique_roots:
        direct_candidates.extend([
            root
            / "data"
            / "kalimantan"
            / "fire_firms"
            / "hourly",

            root
            / "data"
            / "kalimantan"
            / "fire_firms"
            / "daily",

            root
            / "KARHUT_Data_Share"
            / "data"
            / "kalimantan"
            / "fire_firms"
            / "hourly",

            root
            / "KARHUT_Data_Share"
            / "data"
            / "kalimantan"
            / "fire_firms"
            / "daily",
        ])

    # Prefer hourly.
    for candidate in direct_candidates:
        if (
            candidate.name.lower() == "hourly"
            and is_kalimantan_path(candidate)
            and has_firms_files(candidate)
        ):
            return candidate, "hourly"

    # Daily fallback.
    for candidate in direct_candidates:
        if (
            candidate.name.lower() == "daily"
            and is_kalimantan_path(candidate)
            and has_firms_files(candidate)
        ):
            return candidate, "daily"

    # --------------------------------------------------------
    # 2. Recursive fallback, but KALIMANTAN ONLY.
    # --------------------------------------------------------
    matches = []

    for root in unique_roots:
        print("Searching under:", root)

        try:
            for fire_dir in root.rglob("fire_firms"):
                if not fire_dir.is_dir():
                    continue

                # Critical guard: do not select Sumatra.
                if not is_kalimantan_path(fire_dir):
                    continue

                hourly = fire_dir / "hourly"
                daily = fire_dir / "daily"

                if has_firms_files(hourly):
                    matches.append(
                        (hourly, "hourly")
                    )

                if has_firms_files(daily):
                    matches.append(
                        (daily, "daily")
                    )

        except (PermissionError, OSError):
            continue

    if matches:
        matches = sorted(
            matches,
            key=lambda item: (
                0 if item[1] == "hourly" else 1,
                len(str(item[0])),
            ),
        )

        return matches[0]

    return None, None


FIRE_FIRMS_DIR, FIRMS_SOURCE_LEVEL = (
    find_kalimantan_firms_folder(
        DATASET_ROOT
    )
)

if FIRE_FIRMS_DIR is None:
    raise FileNotFoundError(
        "\nCould not locate KALIMANTAN FIRMS files.\n\n"
        "Expected structure similar to:\n"
        "KARHUT_Data_Share/\n"
        "└── data/\n"
        "    └── kalimantan/\n"
        "        └── fire_firms/\n"
        "            └── hourly/\n"
        "                ├── fires_2026-05-01.csv\n"
        "                └── ...\n\n"
        "If the dataset is stored elsewhere, set DATASET_ROOT manually "
        "at the top of this cell."
    )

fire_files = sorted(
    FIRE_FIRMS_DIR.glob(
        "fires_*.csv"
    )
)

if not fire_files:
    raise FileNotFoundError(
        f"No fires_*.csv files found under {FIRE_FIRMS_DIR}"
    )

assert is_kalimantan_path(
    FIRE_FIRMS_DIR
), (
    "Safety check failed: selected FIRMS path is not Kalimantan."
)

print()
print("Raw FIRMS folder :", FIRE_FIRMS_DIR)
print("Source level     :", FIRMS_SOURCE_LEVEL)
print("Files            :", len(fire_files))
print(
    "First / last     :",
    fire_files[0].name,
    "/",
    fire_files[-1].name,
)
````

## 4. Load raw FIRMS detections through the last complete analysis day

The previous pipeline treats 2026-09-11 as partial, so regional selection is based on **2026-05-01 through 2026-09-10**.

````python
# 4. Load raw FIRMS detections

ANALYSIS_START = pd.Timestamp("2026-05-01")
ANALYSIS_END = pd.Timestamp("2026-09-10")

fire_frames = []

date_pattern = re.compile(
    r"fires_(\d{4}-\d{2}-\d{2})\.csv$"
)

for path in fire_files:
    match = date_pattern.search(
        path.name
    )

    if not match:
        continue

    date = pd.Timestamp(
        match.group(1)
    )

    if not (
        ANALYSIS_START
        <= date
        <= ANALYSIS_END
    ):
        continue

    day = pd.read_csv(
        path
    )

    # Empty FIRMS files are valid zero-detection days.
    if day.empty:
        continue

    rename_map = {}

    if (
        "latitude" in day.columns
        and "lat" not in day.columns
    ):
        rename_map["latitude"] = "lat"

    if (
        "longitude" in day.columns
        and "lon" not in day.columns
    ):
        rename_map["longitude"] = "lon"

    day = day.rename(
        columns=rename_map
    )

    required = {
        "lat",
        "lon",
    }

    if not required.issubset(
        day.columns
    ):
        raise ValueError(
            f"{path.name} is missing "
            f"{sorted(required - set(day.columns))}"
        )

    if "frp" not in day.columns:
        day["frp"] = np.nan

    # Hourly files may include time_utc.
    # For this notebook we only need date + position + FRP.
    day = day[
        [
            "lat",
            "lon",
            "frp",
        ]
    ].copy()

    day["date"] = date

    fire_frames.append(
        day
    )

if not fire_frames:
    raise ValueError(
        "No FIRMS rows loaded for the analysis period."
    )

fires_raw = pd.concat(
    fire_frames,
    ignore_index=True,
)

fires_raw["lat"] = pd.to_numeric(
    fires_raw["lat"],
    errors="coerce",
)

fires_raw["lon"] = pd.to_numeric(
    fires_raw["lon"],
    errors="coerce",
)

fires_raw["frp"] = pd.to_numeric(
    fires_raw["frp"],
    errors="coerce",
)

fires_raw = (
    fires_raw
    .dropna(
        subset=[
            "lat",
            "lon",
        ]
    )
    .reset_index(
        drop=True
    )
)

if fires_raw.empty:
    raise ValueError(
        "FIRMS files were found, but no valid coordinates remain after parsing."
    )

lat_min = float(
    fires_raw["lat"].min()
)

lat_max = float(
    fires_raw["lat"].max()
)

lon_min = float(
    fires_raw["lon"].min()
)

lon_max = float(
    fires_raw["lon"].max()
)

print("Raw detections loaded :", len(fires_raw))
print(
    "Date range            :",
    fires_raw["date"].min().date(),
    "→",
    fires_raw["date"].max().date(),
)
print(
    "Unique active days    :",
    fires_raw["date"].nunique(),
)
print(
    "Latitude range        :",
    round(lat_min, 4),
    "→",
    round(lat_max, 4),
)
print(
    "Longitude range       :",
    round(lon_min, 4),
    "→",
    round(lon_max, 4),
)

# Kalimantan package box is around 108–119.5 E.
# This catches accidental selection of Sumatra before geometry filtering.
if lon_max < 108.0 or lon_min > 120.0:
    raise ValueError(
        "Loaded FIRMS coordinates do not look like Kalimantan. "
        f"Longitude range is {lon_min:.3f} → {lon_max:.3f}. "
        "Check DATASET_ROOT / selected FIRMS path."
    )

display(
    fires_raw.head()
)
````

## 5. Download / cache Natural Earth 1:10m layers

Layers:

The download is cached under:

````python
# 5. Natural Earth download helpers

NATURAL_EARTH = {
    "land": (
        "https://naturalearth.s3.amazonaws.com/10m_physical/ne_10m_land.zip",
        "ne_10m_land",
    ),
    "countries": (
        "https://naturalearth.s3.amazonaws.com/10m_cultural/ne_10m_admin_0_countries.zip",
        "ne_10m_admin_0_countries",
    ),
    "admin1": (
        "https://naturalearth.s3.amazonaws.com/10m_cultural/ne_10m_admin_1_states_provinces.zip",
        "ne_10m_admin_1_states_provinces",
    ),
    "lakes": (
        "https://naturalearth.s3.amazonaws.com/10m_physical/ne_10m_lakes.zip",
        "ne_10m_lakes",
    ),
    "rivers": (
        "https://naturalearth.s3.amazonaws.com/10m_physical/ne_10m_rivers_lake_centerlines.zip",
        "ne_10m_rivers_lake_centerlines",
    ),
}


def download_and_extract_ne(
    key,
):
    url, stem = NATURAL_EARTH[key]

    target_dir = (
        EXTERNAL_DIR
        / stem
    )

    shp_path = (
        target_dir
        / f"{stem}.shp"
    )

    if shp_path.exists():
        return shp_path

    target_dir.mkdir(
        parents=True,
        exist_ok=True,
    )

    zip_path = (
        EXTERNAL_DIR
        / f"{stem}.zip"
    )

    if not zip_path.exists():
        print("Downloading:", url)

        response = requests.get(
            url,
            timeout=120,
        )

        response.raise_for_status()

        zip_path.write_bytes(
            response.content
        )

    with zipfile.ZipFile(
        zip_path,
        "r",
    ) as archive:
        archive.extractall(
            target_dir
        )

    if not shp_path.exists():
        alternatives = list(
            target_dir.glob("*.shp")
        )

        if not alternatives:
            raise FileNotFoundError(
                f"No shapefile found after extracting {zip_path.name}"
            )

        shp_path = alternatives[0]

    return shp_path


ne_paths = {
    key: download_and_extract_ne(key)
    for key in NATURAL_EARTH
}

for key, path in ne_paths.items():
    print(f"{key:10s} -> {path}")
````

## 6. Load layers and reconstruct Indonesian Kalimantan

As in the corrected spatial analysis:

1. select the main Borneo land polygon using a seed point around `(114°E, 0.5°N)`;
2. intersect it with Indonesia;
3. use that result as the valid Indonesian Kalimantan land geometry.

````python
# 6. Build Indonesian Kalimantan land geometry

land = gpd.read_file(
    ne_paths["land"]
).to_crs(4326)

countries = gpd.read_file(
    ne_paths["countries"]
).to_crs(4326)

admin1 = gpd.read_file(
    ne_paths["admin1"]
).to_crs(4326)

lakes = gpd.read_file(
    ne_paths["lakes"]
).to_crs(4326)

rivers = gpd.read_file(
    ne_paths["rivers"]
).to_crs(4326)

land_parts = (
    land[["geometry"]]
    .explode(
        index_parts=False
    )
    .reset_index(drop=True)
)

borneo_seed = Point(
    114.0,
    0.5,
)

borneo_candidates = land_parts[
    land_parts.geometry.contains(
        borneo_seed
    )
]

if borneo_candidates.empty:
    raise RuntimeError(
        "Could not identify the main Borneo polygon from Natural Earth."
    )

borneo_geom = (
    borneo_candidates
    .iloc[0]
    .geometry
)

country_mask = pd.Series(
    False,
    index=countries.index,
)

if "ADM0_A3" in countries.columns:
    country_mask |= countries["ADM0_A3"].eq("IDN")

if "SOV_A3" in countries.columns:
    country_mask |= countries["SOV_A3"].eq("IDN")

if "ISO_A2" in countries.columns:
    country_mask |= countries["ISO_A2"].eq("ID")

if "ADMIN" in countries.columns:
    country_mask |= countries["ADMIN"].astype(str).str.lower().eq("indonesia")

indonesia = countries[
    country_mask
].copy()

if indonesia.empty:
    raise RuntimeError(
        "Could not identify Indonesia in Natural Earth admin-0."
    )

indonesia_geom = (
    indonesia.geometry
    .union_all()
)

kalimantan_geom = (
    indonesia_geom
    .intersection(
        borneo_geom
    )
)

kalimantan_gdf = gpd.GeoDataFrame(
    {
        "name": [
            "Indonesian Kalimantan"
        ]
    },
    geometry=[
        kalimantan_geom
    ],
    crs=4326,
)

print(
    "Kalimantan bounds:",
    tuple(
        round(v, 4)
        for v in kalimantan_geom.bounds
    ),
)
````

## 7. Identify Kalimantan admin-1 provinces

Province labels are used only for geographic interpretation of candidate windows.

````python
# 7. Indonesian admin-1 features intersecting Borneo

admin1_indonesia_mask = pd.Series(
    False,
    index=admin1.index,
)

for field, value in [
    ("adm0_a3", "IDN"),
    ("ADM0_A3", "IDN"),
    ("iso_a2", "ID"),
    ("ISO_A2", "ID"),
]:

    if field in admin1.columns:
        admin1_indonesia_mask |= (
            admin1[field]
            .astype(str)
            .eq(value)
        )

if "admin" in admin1.columns:
    admin1_indonesia_mask |= (
        admin1["admin"]
        .astype(str)
        .str.lower()
        .eq("indonesia")
    )

admin1_id = admin1[
    admin1_indonesia_mask
].copy()

admin1_kalimantan = admin1_id[
    admin1_id.geometry.intersects(
        borneo_geom
    )
].copy()

province_name_field = None

for candidate in [
    "name",
    "name_en",
    "gn_name",
    "woe_name",
]:

    if candidate in admin1_kalimantan.columns:
        province_name_field = candidate
        break

if province_name_field is None:
    province_name_field = admin1_kalimantan.columns[0]

province_names = (
    admin1_kalimantan[
        province_name_field
    ]
    .astype(str)
    .tolist()
)

print(
    "Admin-1 regions intersecting Indonesian Borneo:"
)

for name in province_names:
    print("-", name)
````

## 8. Filter raw FIRMS detections to corrected Indonesian Kalimantan geometry

This removes detections outside the corrected land domain before regional selection.

````python
# 8. Geometry filter for raw FIRMS

print(
    "FIRMS coordinate bounds:"
)

print(
    "  lat:",
    round(
        float(
            fires_raw["lat"].min()
        ),
        4,
    ),
    "→",
    round(
        float(
            fires_raw["lat"].max()
        ),
        4,
    ),
)

print(
    "  lon:",
    round(
        float(
            fires_raw["lon"].min()
        ),
        4,
    ),
    "→",
    round(
        float(
            fires_raw["lon"].max()
        ),
        4,
    ),
)

print(
    "Kalimantan geometry bounds:",
    tuple(
        round(
            float(v),
            4,
        )
        for v in kalimantan_geom.bounds
    ),
)

fire_points = gpd.GeoDataFrame(
    fires_raw.copy(),
    geometry=gpd.points_from_xy(
        fires_raw["lon"],
        fires_raw["lat"],
    ),
    crs="EPSG:4326",
)

# For point geometries, intersects() is equivalent to
# inside-or-on-boundary and is vectorized by GeoPandas.
inside_mask = (
    fire_points.geometry
    .intersects(
        kalimantan_geom
    )
)

fires = (
    fire_points.loc[
        inside_mask,
        [
            "date",
            "lat",
            "lon",
            "frp",
            "geometry",
        ],
    ]
    .copy()
    .reset_index(
        drop=True
    )
)

raw_n = len(
    fires_raw
)

inside_n = len(
    fires
)

print()
print(
    "Raw FIRMS rows       :",
    raw_n,
)

print(
    "Inside corrected land:",
    inside_n,
)

print(
    "Retained %           :",
    round(
        100
        * inside_n
        / raw_n,
        2,
    )
    if raw_n
    else 0.0,
)

if fires.empty:
    raise ValueError(
        "Zero FIRMS points remain after the Kalimantan geometry filter. "
        "Compare the printed FIRMS coordinate bounds with "
        "kalimantan_geom.bounds. If they do not overlap, the wrong "
        "regional FIRMS source was loaded."
    )

if inside_n / raw_n < 0.50:
    print(
        "WARNING: less than 50% of loaded FIRMS detections remain. "
        "Inspect the source path and geometry before continuing."
    )
````

# PART C — Discover candidate regional hotspots

## 9. Build a 0.25° raw-detection density surface

This is only for **regional candidate discovery**.

It does not yet define the final modeling grid.

````python
# 9. 0.25-degree density bins

DISCOVERY_STEP = 0.25
ANCHOR_LON = 108.0
ANCHOR_LAT = -4.5

if fires is None or len(fires) == 0:
    raise ValueError(
        "`fires` is empty before density binning. "
        "Rerun and inspect Cell 8 first."
    )

fires = fires.copy()

fires["lon"] = pd.to_numeric(
    fires["lon"],
    errors="coerce",
)

fires["lat"] = pd.to_numeric(
    fires["lat"],
    errors="coerce",
)

fires = (
    fires
    .dropna(
        subset=[
            "lon",
            "lat",
        ]
    )
    .copy()
)

if fires.empty:
    raise ValueError(
        "All FIRMS coordinates became NaN before density binning."
    )

fires["disc_col"] = np.floor(
    (
        fires["lon"]
        - ANCHOR_LON
    )
    / DISCOVERY_STEP
).astype(
    "int64"
)

fires["disc_row"] = np.floor(
    (
        fires["lat"]
        - ANCHOR_LAT
    )
    / DISCOVERY_STEP
).astype(
    "int64"
)

density = (
    fires
    .groupby(
        [
            "disc_row",
            "disc_col",
        ],
        as_index=False,
    )
    .size()
    .rename(
        columns={
            "size": "detections"
        }
    )
)

if density.empty:
    raise ValueError(
        "`density` is empty even though `fires` is non-empty."
    )

# Explicit Python integers are required for np.zeros shape.
row_min = int(
    density["disc_row"].min()
)

row_max = int(
    density["disc_row"].max()
)

col_min = int(
    density["disc_col"].min()
)

col_max = int(
    density["disc_col"].max()
)

n_rows = int(
    row_max
    - row_min
    + 1
)

n_cols = int(
    col_max
    - col_min
    + 1
)

if n_rows <= 0 or n_cols <= 0:
    raise ValueError(
        f"Invalid density matrix dimensions: {n_rows} x {n_cols}"
    )

density_matrix = np.zeros(
    (
        n_rows,
        n_cols,
    ),
    dtype=np.float64,
)

for row in density.itertuples(
    index=False
):
    matrix_row = int(
        row.disc_row
        - row_min
    )

    matrix_col = int(
        row.disc_col
        - col_min
    )

    density_matrix[
        matrix_row,
        matrix_col,
    ] = float(
        row.detections
    )

matrix_total = int(
    density_matrix.sum()
)

print(
    "Grid row range          :",
    row_min,
    "→",
    row_max,
)

print(
    "Grid col range          :",
    col_min,
    "→",
    col_max,
)

print(
    "Discovery density matrix:",
    density_matrix.shape,
)

print(
    "Occupied 0.25° cells    :",
    len(
        density
    ),
)

print(
    "Total detections         :",
    matrix_total,
)

print(
    "Original FIRMS rows      :",
    len(
        fires
    ),
)

assert matrix_total == len(
    fires
), (
    "Density-matrix total does not equal number of FIRMS detections."
)
````

## 10. Smooth density using a local 5×5 neighborhood

At 0.25°, a 5×5 neighborhood represents roughly a 1.25° regional activity window.

No SciPy dependency is required.

````python
# 10. Simple 5x5 neighborhood sum

def box_sum_2d(
    matrix,
    radius_cells=2,
):

    padded = np.pad(
        matrix,
        pad_width=radius_cells,
        mode="constant",
        constant_values=0,
    )

    output = np.zeros_like(
        matrix,
        dtype=float,
    )

    width = (
        2
        * radius_cells
        + 1
    )

    for dr in range(width):
        for dc in range(width):
            output += padded[
                dr:
                dr + matrix.shape[0],
                dc:
                dc + matrix.shape[1],
            ]

    return output


smoothed_density = box_sum_2d(
    density_matrix,
    radius_cells=2,
)

print(
    "Peak local 5x5 detections:",
    int(
        smoothed_density.max()
    ),
)
````

## 11. Select non-overlapping candidate centers

Candidate window size:

```text
3° × 3°
```

This is approximately a few hundred kilometers across and is large enough to contain a meaningful regional graph while still being much more specific than all Kalimantan.

Top peaks are forced to be geographically separated so we do not return several almost-identical boxes around one cluster.

````python
# 11. Candidate peak selection

N_CANDIDATES = 5
MIN_CENTER_SEPARATION_DEG = 2.0
WINDOW_WIDTH_DEG = 3.0
HALF_WIDTH = WINDOW_WIDTH_DEG / 2.0

flat_order = np.argsort(
    smoothed_density.ravel()
)[::-1]

candidate_centers = []

for flat_index in flat_order:

    matrix_row, matrix_col = np.unravel_index(
        flat_index,
        smoothed_density.shape,
    )

    disc_row = (
        matrix_row
        + row_min
    )

    disc_col = (
        matrix_col
        + col_min
    )

    center_lat = (
        ANCHOR_LAT
        + (
            disc_row
            + 0.5
        )
        * DISCOVERY_STEP
    )

    center_lon = (
        ANCHOR_LON
        + (
            disc_col
            + 0.5
        )
        * DISCOVERY_STEP
    )

    center_point = Point(
        center_lon,
        center_lat,
    )

    if not center_point.within(
        kalimantan_geom
    ):
        continue

    too_close = False

    for previous in candidate_centers:

        distance_deg = math.hypot(
            center_lon
            - previous[
                "center_lon"
            ],
            center_lat
            - previous[
                "center_lat"
            ],
        )

        if (
            distance_deg
            < MIN_CENTER_SEPARATION_DEG
        ):
            too_close = True
            break

    if too_close:
        continue

    candidate_centers.append({
        "center_lon": center_lon,
        "center_lat": center_lat,
        "peak_density_score": float(
            smoothed_density[
                matrix_row,
                matrix_col,
            ]
        ),
    })

    if (
        len(candidate_centers)
        >= N_CANDIDATES
    ):
        break


candidates = pd.DataFrame(
    candidate_centers
)

candidates["candidate_id"] = [
    f"C{i+1}"
    for i in range(
        len(candidates)
    )
]

candidates["west"] = (
    candidates["center_lon"]
    - HALF_WIDTH
)

candidates["east"] = (
    candidates["center_lon"]
    + HALF_WIDTH
)

candidates["south"] = (
    candidates["center_lat"]
    - HALF_WIDTH
)

candidates["north"] = (
    candidates["center_lat"]
    + HALF_WIDTH
)

display(
    candidates[
        [
            "candidate_id",
            "center_lon",
            "center_lat",
            "peak_density_score",
            "west",
            "south",
            "east",
            "north",
        ]
    ]
)
````

# PART D — Fine-grid event diagnostic for candidate ranking

## 12. Helper: valid grid centers inside Indonesian Kalimantan

Candidate ranking uses a **0.25° preview grid**.

A grid cell is considered for the ranking if its center falls inside the corrected Indonesian Kalimantan land geometry.

Notebook 17 will perform a more rigorous overlap-based land/water mask.

````python
# 12. Grid-center helper

def aligned_index_range(
    low,
    high,
    anchor,
    step,
):

    start = math.floor(
        (
            low
            - anchor
        )
        / step
    )

    end = math.floor(
        (
            high
            - anchor
        )
        / step
    )

    return range(
        start,
        end + 1,
    )


def valid_grid_centers(
    bbox_tuple,
    step,
):

    west, south, east, north = (
        bbox_tuple
    )

    rows = []

    for col_idx in aligned_index_range(
        west,
        east,
        ANCHOR_LON,
        step,
    ):

        cell_west = (
            ANCHOR_LON
            + col_idx
            * step
        )

        center_lon = (
            cell_west
            + step / 2
        )

        if not (
            west
            <= center_lon
            <= east
        ):
            continue

        for row_idx in aligned_index_range(
            south,
            north,
            ANCHOR_LAT,
            step,
        ):

            cell_south = (
                ANCHOR_LAT
                + row_idx
                * step
            )

            center_lat = (
                cell_south
                + step / 2
            )

            if not (
                south
                <= center_lat
                <= north
            ):
                continue

            center = Point(
                center_lon,
                center_lat,
            )

            if center.within(
                kalimantan_geom
            ):

                rows.append({
                    "row_idx": row_idx,
                    "col_idx": col_idx,
                    "center_lat": center_lat,
                    "center_lon": center_lon,
                    "cell_id": (
                        f"g{step:.2f}_"
                        f"{row_idx}_"
                        f"{col_idx}"
                    ),
                })

    return pd.DataFrame(
        rows
    )
````

## 13. Helper: build 3-day Emerging panel

Definition:

```text
Emerging at target day t+1
=
target cell active at t+1
AND
target cell had zero detected activity during t-2, t-1, t
```

As before, this means **emerging FIRMS detection**, not proven new fire ignition.

````python
# 13. Candidate fine-grid panel statistics

ALL_DATES = pd.date_range(
    ANALYSIS_START,
    ANALYSIS_END,
    freq="D",
)


def candidate_event_stats(
    fire_df,
    bbox_tuple,
    step=0.25,
):

    west, south, east, north = (
        bbox_tuple
    )

    valid_cells = valid_grid_centers(
        bbox_tuple,
        step,
    )

    if valid_cells.empty:
        return {
            "grid_step": step,
            "valid_cells": 0,
            "active_cells": 0,
            "positive_cell_days": 0,
            "emerging3": 0,
            "persistent3": 0,
            "target_rows": 0,
            "target_positive_rate": np.nan,
        }

    regional = fire_df[
        fire_df["lon"].between(
            west,
            east,
            inclusive="both",
        )
        &
        fire_df["lat"].between(
            south,
            north,
            inclusive="both",
        )
    ].copy()

    regional["col_idx"] = np.floor(
        (
            regional["lon"]
            - ANCHOR_LON
        )
        / step
    ).astype(int)

    regional["row_idx"] = np.floor(
        (
            regional["lat"]
            - ANCHOR_LAT
        )
        / step
    ).astype(int)

    lookup = valid_cells[
        [
            "row_idx",
            "col_idx",
            "cell_id",
        ]
    ]

    regional = regional.merge(
        lookup,
        on=[
            "row_idx",
            "col_idx",
        ],
        how="inner",
    )

    active_daily = (
        regional
        .groupby(
            [
                "date",
                "cell_id",
            ],
            as_index=False,
        )
        .agg(
            fire_count=(
                "frp",
                "size",
            ),
            frp_sum=(
                "frp",
                "sum",
            ),
        )
    )

    full_index = pd.MultiIndex.from_product(
        [
            ALL_DATES,
            valid_cells[
                "cell_id"
            ],
        ],
        names=[
            "date",
            "cell_id",
        ],
    )

    panel = (
        active_daily
        .set_index(
            [
                "date",
                "cell_id",
            ]
        )
        .reindex(
            full_index
        )
        .reset_index()
    )

    panel[
        [
            "fire_count",
            "frp_sum",
        ]
    ] = (
        panel[
            [
                "fire_count",
                "frp_sum",
            ]
        ]
        .fillna(0)
    )

    panel["fire_active"] = (
        panel["fire_count"]
        > 0
    ).astype(int)

    panel = panel.sort_values(
        [
            "cell_id",
            "date",
        ]
    ).reset_index(drop=True)

    panel["recent3_incl_t"] = (
        panel
        .groupby(
            "cell_id"
        )[
            "fire_active"
        ]
        .transform(
            lambda s: (
                s.rolling(
                    3,
                    min_periods=3,
                )
                .sum()
            )
        )
    )

    panel["target_t1"] = (
        panel
        .groupby(
            "cell_id"
        )[
            "fire_active"
        ]
        .shift(-1)
    )

    model_panel = panel[
        panel["recent3_incl_t"].notna()
        &
        panel["target_t1"].notna()
    ].copy()

    model_panel["emerging3"] = (
        model_panel["target_t1"].eq(1)
        &
        model_panel["recent3_incl_t"].eq(0)
    )

    model_panel["persistent3"] = (
        model_panel["target_t1"].eq(1)
        &
        model_panel["recent3_incl_t"].gt(0)
    )

    return {
        "grid_step": step,
        "valid_cells": int(
            len(
                valid_cells
            )
        ),
        "active_cells": int(
            regional[
                "cell_id"
            ].nunique()
        ),
        "positive_cell_days": int(
            panel[
                "fire_active"
            ].sum()
        ),
        "emerging3": int(
            model_panel[
                "emerging3"
            ].sum()
        ),
        "persistent3": int(
            model_panel[
                "persistent3"
            ].sum()
        ),
        "target_rows": int(
            len(
                model_panel
            )
        ),
        "target_positive_rate": float(
            model_panel[
                "target_t1"
            ].mean()
        ),
    }
````

## 14. Compute candidate activity + 0.25° Emerging statistics

````python
# 14. Candidate statistics

candidate_rows = []

for row in candidates.itertuples(
    index=False
):

    bbox_tuple = (
        row.west,
        row.south,
        row.east,
        row.north,
    )

    regional = fires[
        fires["lon"].between(
            row.west,
            row.east,
            inclusive="both",
        )
        &
        fires["lat"].between(
            row.south,
            row.north,
            inclusive="both",
        )
    ].copy()

    event_stats = (
        candidate_event_stats(
            fires,
            bbox_tuple,
            step=0.25,
        )
    )

    candidate_rows.append({
        "candidate_id": row.candidate_id,
        "center_lon": row.center_lon,
        "center_lat": row.center_lat,
        "west": row.west,
        "south": row.south,
        "east": row.east,
        "north": row.north,
        "peak_density_score": row.peak_density_score,
        "detections": int(
            len(
                regional
            )
        ),
        "active_days": int(
            regional[
                "date"
            ].nunique()
        ),
        "frp_sum": float(
            regional[
                "frp"
            ].sum(
                skipna=True
            )
        ),
        "grid025_valid_cells": event_stats[
            "valid_cells"
        ],
        "grid025_active_cells": event_stats[
            "active_cells"
        ],
        "grid025_positive_cell_days": event_stats[
            "positive_cell_days"
        ],
        "grid025_emerging3": event_stats[
            "emerging3"
        ],
        "grid025_persistent3": event_stats[
            "persistent3"
        ],
        "grid025_target_positive_rate": event_stats[
            "target_positive_rate"
        ],
    })


candidate_stats = pd.DataFrame(
    candidate_rows
)

display(
    candidate_stats
)
````

## 16. Candidate province names, lake area, and river length

Hydrography metrics are descriptive only.

They are **not included in candidate ranking**.

````python
# 16. Candidate hydrography context

candidate_context_rows = []

for row in candidate_stats.itertuples(
    index=False
):

    bbox_geom = box(
        row.west,
        row.south,
        row.east,
        row.north,
    )

    province_hits = (
        admin1_kalimantan[
            admin1_kalimantan.geometry.intersects(
                bbox_geom
            )
        ][
            province_name_field
        ]
        .astype(str)
        .tolist()
    )

    candidate_lakes = lakes_kalimantan[
        lakes_kalimantan.geometry.intersects(
            bbox_geom
        )
    ].copy()

    candidate_rivers = rivers_kalimantan[
        rivers_kalimantan.geometry.intersects(
            bbox_geom
        )
    ].copy()

    if candidate_lakes.empty:
        lake_area_km2 = 0.0
    else:
        lake_clipped = candidate_lakes.copy()
        lake_clipped["geometry"] = (
            lake_clipped.geometry.intersection(
                bbox_geom
            )
        )

        lake_area_km2 = float(
            lake_clipped
            .to_crs(6933)
            .geometry
            .area
            .sum()
            / 1e6
        )

    if candidate_rivers.empty:
        river_length_km = 0.0
    else:
        river_clipped = candidate_rivers.copy()
        river_clipped["geometry"] = (
            river_clipped.geometry.intersection(
                bbox_geom
            )
        )

        river_length_km = float(
            river_clipped
            .to_crs(32650)
            .geometry
            .length
            .sum()
            / 1000
        )

    candidate_context_rows.append({
        "candidate_id": row.candidate_id,
        "provinces": " | ".join(
            sorted(
                set(
                    province_hits
                )
            )
        ),
        "natural_earth_lakes_n": int(
            len(
                candidate_lakes
            )
        ),
        "natural_earth_lake_area_km2": lake_area_km2,
        "natural_earth_river_features_n": int(
            len(
                candidate_rivers
            )
        ),
        "natural_earth_river_length_km": river_length_km,
    })


candidate_context = pd.DataFrame(
    candidate_context_rows
)

candidate_table = candidate_stats.merge(
    candidate_context,
    on="candidate_id",
    how="left",
    validate="one_to_one",
)

display(
    candidate_table
)
````

## 17. Ranking rule

````python
# 17. Data-driven ranking

ranked_candidates = (
    candidate_table
    .sort_values(
        [
            "grid025_emerging3",
            "active_days",
            "grid025_active_cells",
            "detections",
        ],
        ascending=[
            False,
            False,
            False,
            False,
        ],
    )
    .reset_index(drop=True)
)

ranked_candidates[
    "selection_rank"
] = (
    np.arange(
        len(
            ranked_candidates
        )
    )
    + 1
)

ranked_candidates = ranked_candidates[
    [
        "selection_rank",
        *[
            c
            for c in ranked_candidates.columns
            if c != "selection_rank"
        ],
    ]
]

display(
    ranked_candidates
)
````

# PART I — Build rigorous land/water preview grids

## 23. Fine-grid polygon helper

For every cell we calculate:

- cell center;
- `land_fraction` = overlap with corrected Indonesian Kalimantan land;
- `water_fraction = 1 - land_fraction`;
- `lake_fraction` = overlap with Natural Earth lake polygons;
- `river_intersects` = whether a mapped Natural Earth river crosses the cell.

`water_fraction` captures coastline/ocean and large non-land areas in the Natural Earth geometry. `lake_fraction` is reported separately.

For final thesis modeling, do not treat `river_intersects=True` as “this whole cell is water”; rivers are lines and most cells remain predominantly land.

````python
# 23. Land/water fine-grid builder

def build_landwater_grid(
    bbox_tuple,
    step,
    min_land_fraction=0.05,
):

    west, south, east, north = (
        bbox_tuple
    )

    rows = []

    for col_idx in aligned_index_range(
        west,
        east,
        ANCHOR_LON,
        step,
    ):

        cell_west = (
            ANCHOR_LON
            + col_idx
            * step
        )

        cell_east = (
            cell_west
            + step
        )

        if (
            cell_east <= west
            or cell_west >= east
        ):
            continue

        for row_idx in aligned_index_range(
            south,
            north,
            ANCHOR_LAT,
            step,
        ):

            cell_south = (
                ANCHOR_LAT
                + row_idx
                * step
            )

            cell_north = (
                cell_south
                + step
            )

            if (
                cell_north <= south
                or cell_south >= north
            ):
                continue

            geom = box(
                cell_west,
                cell_south,
                cell_east,
                cell_north,
            )

            rows.append({
                "row_idx": row_idx,
                "col_idx": col_idx,
                "cell_id": (
                    f"g{step:.2f}_"
                    f"{row_idx}_"
                    f"{col_idx}"
                ),
                "center_lon": (
                    cell_west
                    + step / 2
                ),
                "center_lat": (
                    cell_south
                    + step / 2
                ),
                "geometry": geom,
            })

    grid = gpd.GeoDataFrame(
        rows,
        geometry="geometry",
        crs=4326,
    )

    if grid.empty:
        return grid

    grid_equal = grid.to_crs(
        6933
    )

    kalimantan_equal = (
        kalimantan_gdf
        .to_crs(
            6933
        )
        .geometry
        .iloc[0]
    )

    grid["cell_area_m2"] = (
        grid_equal.geometry.area
        .to_numpy()
    )

    land_intersection_area = (
        grid_equal.geometry
        .intersection(
            kalimantan_equal
        )
        .area
        .to_numpy()
    )

    grid["land_fraction"] = (
        land_intersection_area
        / grid["cell_area_m2"]
    )

    grid["water_fraction"] = (
        1.0
        - grid["land_fraction"]
    ).clip(
        0.0,
        1.0,
    )

    if chosen_lakes.empty:

        grid["lake_fraction"] = 0.0

    else:

        lakes_union_equal = (
            chosen_lakes
            .to_crs(
                6933
            )
            .geometry
            .union_all()
        )

        lake_intersection_area = (
            grid_equal.geometry
            .intersection(
                lakes_union_equal
            )
            .area
            .to_numpy()
        )

        grid["lake_fraction"] = (
            lake_intersection_area
            / grid["cell_area_m2"]
        ).clip(
            0.0,
            1.0,
        )

    if chosen_rivers.empty:

        grid["river_intersects"] = False

    else:

        rivers_union = (
            chosen_rivers.geometry
            .union_all()
        )

        grid["river_intersects"] = (
            grid.geometry.intersects(
                rivers_union
            )
        )

    center_points = gpd.GeoSeries(
        gpd.points_from_xy(
            grid["center_lon"],
            grid["center_lat"],
        ),
        crs=4326,
    )

    grid["center_inside_land"] = (
        center_points.within(
            kalimantan_geom
        )
        .to_numpy()
    )

    grid["keep_candidate"] = (
        grid["land_fraction"].ge(
            min_land_fraction
        )
        |
        grid["center_inside_land"]
    )

    grid["land_dominant"] = (
        grid["land_fraction"]
        >= 0.50
    )

    grid["water_dominant"] = (
        grid["land_fraction"]
        < 0.50
    )

    grid["lake_dominant"] = (
        grid["lake_fraction"]
        >= 0.50
    )

    return grid
````

## 24. Compare 0.25° and 0.10° spatial realism

````python
# 24. Resolution grid summaries

resolution_grids = {}
resolution_summary_rows = []

for step in [
    0.25,
    0.10,
]:

    grid = build_landwater_grid(
        CHOSEN_BBOX,
        step,
        min_land_fraction=0.05,
    )

    resolution_grids[
        step
    ] = grid

    kept = grid[
        grid[
            "keep_candidate"
        ]
    ]

    resolution_summary_rows.append({
        "Grid Step": step,
        "All Rectangular Cells": len(
            grid
        ),
        "Candidate Cells After Land Filter": len(
            kept
        ),
        "Land-Dominant Cells": int(
            kept[
                "land_dominant"
            ].sum()
        ),
        "Water-Dominant Mixed Cells": int(
            kept[
                "water_dominant"
            ].sum()
        ),
        "Lake-Dominant Cells": int(
            kept[
                "lake_dominant"
            ].sum()
        ),
        "Cells Crossed by Mapped River": int(
            kept[
                "river_intersects"
            ].sum()
        ),
        "Median Land Fraction": float(
            kept[
                "land_fraction"
            ].median()
        ),
    })


resolution_geo_summary = pd.DataFrame(
    resolution_summary_rows
)

display(
    resolution_geo_summary
)
````
